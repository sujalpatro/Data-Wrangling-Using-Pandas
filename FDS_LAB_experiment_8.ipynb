```python
import pandas as pd

# --------------------------------------------------
# 1. CREATE DATA FRAMES
# --------------------------------------------------

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Sujal', 'Rahul', 'Aman', 'Priya'],
    'Department': ['CSE', 'ECE', 'CSE', 'EEE'],
    'Marks': [85, 78, 92, 88]
})

df2 = pd.DataFrame({
    'ID': [5, 6, 7, 8],
    'Name': ['Riya', 'Arjun', 'Kiran', 'Neha'],
    'Department': ['CSE', 'ECE', 'EEE', 'CSE'],
    'Marks': [90, 75, 82, 95]
})

print("DATA FRAME 1")
print(df1)

print("\nDATA FRAME 2")
print(df2)


# --------------------------------------------------
# 2. HIERARCHICAL INDEXING / MULTIINDEX
# --------------------------------------------------

df_multi = df1.set_index(['Department', 'ID'])

print("\nHIERARCHICAL INDEXING")
print(df_multi)

# Access CSE students
print("\nCSE STUDENTS")
print(df_multi.loc['CSE'])


# --------------------------------------------------
# 3. COMBINING DATAFRAMES USING CONCAT
# --------------------------------------------------

combined_df = pd.concat([df1, df2], ignore_index=True)

print("\nCOMBINED DATAFRAME")
print(combined_df)


# --------------------------------------------------
# 4. MERGING DATAFRAMES
# --------------------------------------------------

student = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Sujal', 'Rahul', 'Aman', 'Priya']
})

result = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'CGPA': [8.5, 7.8, 9.2, 8.8]
})

merged_df = pd.merge(student, result, on='ID')

print("\nMERGED DATAFRAME")
print(merged_df)


# --------------------------------------------------
# 5. RESHAPING USING STACK
# --------------------------------------------------

marks = pd.DataFrame({
    'Name': ['Sujal', 'Rahul', 'Aman'],
    'Math': [85, 78, 92],
    'Python': [90, 80, 88]
})

print("\nORIGINAL MARKS DATAFRAME")
print(marks)

stacked = marks.set_index('Name').stack()

print("\nSTACKED DATAFRAME")
print(stacked)


# --------------------------------------------------
# 6. RESHAPING USING UNSTACK
# --------------------------------------------------

unstacked = stacked.unstack()

print("\nUNSTACKED DATAFRAME")
print(unstacked)


# --------------------------------------------------
# 7. PIVOTING
# --------------------------------------------------

sales = pd.DataFrame({
    'Name': ['Sujal', 'Rahul', 'Sujal', 'Rahul'],
    'Subject': ['Python', 'Python', 'DBMS', 'DBMS'],
    'Marks': [90, 80, 85, 75]
})

print("\nSALES / MARKS DATA")
print(sales)

pivot_df = sales.pivot(
    index='Name',
    columns='Subject',
    values='Marks'
)

print("\nPIVOTED DATAFRAME")
print(pivot_df)


# --------------------------------------------------
# 8. PIVOT TABLE
# --------------------------------------------------

pivot_table_df = sales.pivot_table(
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='mean'
)

print("\nPIVOT TABLE")
print(pivot_table_df)


# --------------------------------------------------
# 9. MELT - WIDE TO LONG FORMAT
# --------------------------------------------------

melted_df = pivot_df.reset_index().melt(
    id_vars='Name',
    var_name='Subject',
    value_name='Marks'
)

print("\nMELTED DATAFRAME")
print(melted_df)
```
